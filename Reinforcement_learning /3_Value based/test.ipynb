**Политика и награда (более строго)** — означает, что речь пойдёт о математически точных определениях этих ключевых понятий.

---

### Что такое политика?

> **Политикой** называется отображение из пространства состояний $ \mathcal{S} $ в пространство действий $ \mathcal{A} $:

$
\pi(a|s): \mathcal{S} \mapsto \mathcal{A}, \quad \pi(a|s) \in [0, 1]
$

- $ \pi(a|s) $ — вероятность того, что агент выберет действие $ a $, находясь в состоянии $ s $.
- $ a \in \mathcal{A} $ — возможные действия.
- $ s \in \mathcal{S} $ — возможные состояния среды.
- Это определение описывает **стохастическую политику** (если вероятности), но можно также рассматривать детерминированную политику, где $ \pi(s) = a $.

---

### Что делает политика?

Результатом применения политики в среде является **траектория** (или эпизод):

$
\tau^\pi = \{S_0, A_0, S_1, A_1, \ldots, S_F\}
$

- Это последовательность чередующихся состояний $ S_t $ и действий $ A_t $, получающаяся при следовании политике $ \pi $.
- $ F $ — финальное состояние эпизода (может быть конечным или бесконечным в зависимости от задачи).

---

### Награда за траекторию:

$
G(\tau) = \sum_{t=0}^{\infty} \gamma^t R(S_t, A_t)
$

- $ R(S_t, A_t) $ — вознаграждение, полученное на шаге $ t $.
- $ \gamma \in [0, 1] $ — **коэффициент дисконтирования**: уменьшает влияние наград в далёком будущем.
- $ G(\tau) $ — суммарная (дисконтированная) награда за всю траекторию $ \tau $.

---

### Цель агента:

> **Максимизировать суммарное вознаграждение**, т.е. найти такую политику $ \pi $, которая максимизирует:

$
G(\tau^\pi) = \sum_{t=0}^{\infty} \gamma^t R_t
$

Здесь то же самое, но краткая форма записи (слева она выделена красной рамкой).

---

### Подытожим:

- Агент живёт в среде, в каждом состоянии выбирает действие согласно политике.
- Эти действия приводят к новой ситуации и вознаграждению.
- Агент хочет максимизировать сумму всех наград (с учётом уменьшения важности будущих наград через $ \gamma $).
- Задача обучения с подкреплением — **найти оптимальную политику** $ \pi^* $, которая делает это лучше всего.
